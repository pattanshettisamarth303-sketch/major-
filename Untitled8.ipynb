```python
THRESHOLDS = {
    "N": {"low": 280, "high": 560},       # kg/ha
    "P": {"low": 10, "high": 25},         # kg/ha
    "K": {"low": 108, "high": 280},       # kg/ha
    "pH": {"acidic": 6.0, "alkaline": 7.5},
    "organic_matter": {"low": 0.50, "high": 0.75}  # percentage
}
```

```python
KNOWLEDGE_BASE = {
    "N": {
        "low": {
            "status": "Nitrogen Deficient",
            "fertilizer": "Urea (46% N) or Ammonium Sulfate",
            "organic": "Compost, Farmyard Manure (FYM), or Green Manure",
            "action": "Apply nitrogenous fertilizers in split doses to minimize leaching."
        },
        "high": {
            "status": "Excess Nitrogen",
            "fertilizer": "None",
            "organic": "None",
            "action": "Skip nitrogen application; excessive N promotes vegetative growth over yield."
        }
    },
    "P": {
        "low": {
            "status": "Phosphorus Deficient",
            "fertilizer": "Single Super Phosphate (SSP) or DAP",
            "organic": "Bone meal or Rock Phosphate",
            "action": "Apply phosphorus at seed depth during sowing for optimal root development."
        },
        "high": {
            "status": "Excess Phosphorus",
            "fertilizer": "None",
            "organic": "None",
            "action": "Omit P application; excess phosphorus can lock micronutrients like Zinc."
        }
    },
    "K": {
        "low": {
            "status": "Potassium Deficient",
            "fertilizer": "Muriate of Potash (MOP) or Potassium Sulfate",
            "organic": "Wood ash or composted seaweed",
            "action": "Incorporate potassium into soil before planting to improve stress tolerance."
        },
        "high": {
            "status": "Excess Potassium",
            "fertilizer": "None",
            "organic": "None",
            "action": "Omit K application; high levels can restrict Magnesium and Calcium uptake."
        }
    },
    "pH": {
        "acidic": {
            "status": "Acidic Soil",
            "amendment": "Agricultural Lime (Calcium Carbonate)",
            "action": "Apply lime 2–4 weeks before planting to elevate soil pH to neutral levels."
        },
        "alkaline": {
            "status": "Alkaline Soil",
            "amendment": "Gypsum (Calcium Sulfate) or Elemental Sulfur",
            "action": "Apply gypsum/sulfur to reduce alkalinity and enhance nutrient availability."
        }
    },
    "organic_matter": {
        "low": {
            "status": "Low Organic Matter",
            "amendment": "Vermicompost or Well-rotted Farmyard Manure",
            "action": "Apply 5–10 tons/hectare of organic matter to boost microbial activity and water retention."
        }
    }
}
```

```python
def generate_step1_recommendations(n, p, k, ph, organic_matter):
    recommendations = {
        "nutrient_status": {},
        "fertilizer_recommendations": [],
        "soil_amendments": [],
        "application_guidance": []
    }

    # Nitrogen Rules
    if n < THRESHOLDS["N"]["low"]:
        recommendations["nutrient_status"]["N"] = "Low"
        recommendations["fertilizer_recommendations"].append(KNOWLEDGE_BASE["N"]["low"])
    elif n > THRESHOLDS["N"]["high"]:
        recommendations["nutrient_status"]["N"] = "High"
        recommendations["fertilizer_recommendations"].append(KNOWLEDGE_BASE["N"]["high"])
    else:
        recommendations["nutrient_status"]["N"] = "Optimal"

    # Phosphorus Rules
    if p < THRESHOLDS["P"]["low"]:
        recommendations["nutrient_status"]["P"] = "Low"
        recommendations["fertilizer_recommendations"].append(KNOWLEDGE_BASE["P"]["low"])
    elif p > THRESHOLDS["P"]["high"]:
        recommendations["nutrient_status"]["P"] = "High"
        recommendations["fertilizer_recommendations"].append(KNOWLEDGE_BASE["P"]["high"])
    else:
        recommendations["nutrient_status"]["P"] = "Optimal"

    # Potassium Rules
    if k < THRESHOLDS["K"]["low"]:
        recommendations["nutrient_status"]["K"] = "Low"
        recommendations["fertilizer_recommendations"].append(KNOWLEDGE_BASE["K"]["low"])
    elif k > THRESHOLDS["K"]["high"]:
        recommendations["nutrient_status"]["K"] = "High"
        recommendations["fertilizer_recommendations"].append(KNOWLEDGE_BASE["K"]["high"])
    else:
        recommendations["nutrient_status"]["K"] = "Optimal"

    # pH Rules
    if ph < THRESHOLDS["pH"]["acidic"]:
        recommendations["nutrient_status"]["pH"] = "Acidic"
        recommendations["soil_amendments"].append(KNOWLEDGE_BASE["pH"]["acidic"])
    elif ph > THRESHOLDS["pH"]["alkaline"]:
        recommendations["nutrient_status"]["pH"] = "Alkaline"
        recommendations["soil_amendments"].append(KNOWLEDGE_BASE["pH"]["alkaline"])
    else:
        recommendations["nutrient_status"]["pH"] = "Neutral (Optimal)"

    # Organic Matter Rules
    if organic_matter < THRESHOLDS["organic_matter"]["low"]:
        recommendations["nutrient_status"]["Organic Matter"] = "Low"
        recommendations["soil_amendments"].append(KNOWLEDGE_BASE["organic_matter"]["low"])
    else:
        recommendations["nutrient_status"]["Organic Matter"] = "Adequate"

    # Application Guidance
    recommendations["application_guidance"] = [
        "Basal Application: Apply 100% of P, 100% of K, and 50% of required N during land preparation.",
        "Top Dressing: Apply the remaining 50% of N in split applications during peak vegetative stages.",
        "Safety Rule: Maintain adequate moisture during fertilizer application to avoid crop root burn."
    ]

    return recommendations
```

```python
import json

sample_soil = {
    "n": 180,
    "p": 8,
    "k": 200,
    "ph": 5.2,
    "organic_matter": 0.35
}

result = generate_step1_recommendations(**sample_soil)
print(json.dumps(result, indent=2))
```

```python
class RecommendationEngine:
    def __init__(self, thresholds, knowledge_base):
        self.thresholds = thresholds
        self.knowledge_base = knowledge_base

    def evaluate_soil(self, soil_sample):
        """Processes soil parameters and returns structured recommendations."""
        n = soil_sample.get("n", 0)
        p = soil_sample.get("p", 0)
        k = soil_sample.get("k", 0)
        ph = soil_sample.get("ph", 7.0)
        om = soil_sample.get("organic_matter", 1.0)

        # Generate base recommendations
        return generate_step1_recommendations(n, p, k, ph, om)
```

```python
# Initialize Engine
engine = RecommendationEngine(THRESHOLDS, KNOWLEDGE_BASE)

# Define test datasets
test_cases = {
    "Sample A (Severely Deficient & Acidic)": {
        "n": 150, "p": 5, "k": 90, "ph": 5.0, "organic_matter": 0.3
    },
    "Sample B (Alkaline & High Nitrogen)": {
        "n": 600, "p": 15, "k": 200, "ph": 8.2, "organic_matter": 0.8
    },
    "Sample C (Optimal Balanced Soil)": {
        "n": 350, "p": 18, "k": 220, "ph": 6.8, "organic_matter": 0.8
    }
}

# Run tests
for name, sample in test_cases.items():
    print(f"=== {name} ===")
    output = engine.evaluate_soil(sample)
    print("Nutrient Status:", output["nutrient_status"])
    print("Fertilizers Needed:", len(output["fertilizer_recommendations"]))
    print("Amendments Needed:", len(output["soil_amendments"]))
    print("-" * 40)
```

```python
# Crop Database with Optimal Parameter Ranges
CROP_DATABASE = [
    {
        "crop": "Wheat",
        "n_min": 250, "n_max": 400,
        "p_min": 15,  "p_max": 30,
        "k_min": 150, "k_max": 250,
        "ph_min": 6.0, "ph_max": 7.5
    },
    {
        "crop": "Rice (Paddy)",
        "n_min": 280, "n_max": 450,
        "p_min": 20,  "p_max": 35,
        "k_min": 180, "k_max": 300,
        "ph_min": 5.5, "ph_max": 6.8
    },
    {
        "crop": "Maize (Corn)",
        "n_min": 300, "n_max": 500,
        "p_min": 25,  "p_max": 45,
        "k_min": 160, "k_max": 280,
        "ph_min": 5.8, "ph_max": 7.2
    },
    {
        "crop": "Cotton",
        "n_min": 200, "n_max": 350,
        "p_min": 10,  "p_max": 25,
        "k_min": 120, "k_max": 220,
        "ph_min": 6.5, "ph_max": 8.0
    },
    {
        "crop": "Potato",
        "n_min": 180, "n_max": 320,
        "p_min": 30,  "p_max": 50,
        "k_min": 200, "k_max": 350,
        "ph_min": 5.0, "ph_max": 6.5
    }
]
```

```python
def calculate_crop_suitability(soil, crop_db):
    ranked_crops = []

    for crop in crop_db:
        score = 100.0
        penalties = []

        # Evaluate N match
        if not (crop["n_min"] <= soil["n"] <= crop["n_max"]):
            diff = min(abs(soil["n"] - crop["n_min"]), abs(soil["n"] - crop["n_max"]))
            penalty = min(25, (diff / crop["n_min"]) * 20)
            score -= penalty
            penalties.append(f"Nitrogen deviation")

        # Evaluate P match
        if not (crop["p_min"] <= soil["p"] <= crop["p_max"]):
            diff = min(abs(soil["p"] - crop["p_min"]), abs(soil["p"] - crop["p_max"]))
            penalty = min(25, (diff / crop["p_min"]) * 20)
            score -= penalty
            penalties.append(f"Phosphorus deviation")

        # Evaluate K match
        if not (crop["k_min"] <= soil["k"] <= crop["k_max"]):
            diff = min(abs(soil["k"] - crop["k_min"]), abs(soil["k"] - crop["k_max"]))
            penalty = min(25, (diff / crop["k_min"]) * 20)
            score -= penalty
            penalties.append(f"Potassium deviation")

        # Evaluate pH match
        if not (crop["ph_min"] <= soil["ph"] <= crop["ph_max"]):
            diff = min(abs(soil["ph"] - crop["ph_min"]), abs(soil["ph"] - crop["ph_max"]))
            penalty = min(25, diff * 15)
            score -= penalty
            penalties.append(f"pH level non-optimal")

        final_score = max(0.0, round(score, 2))

        # Categorize Suitability Level
        if final_score >= 85:
            suitability = "Highly Suitable"
        elif final_score >= 65:
            suitability = "Moderately Suitable"
        elif final_score >= 45:
            suitability = "Marginally Suitable"
        else:
            suitability = "Not Recommended"

        ranked_crops.append({
            "crop": crop["crop"],
            "score": final_score,
            "suitability": suitability,
            "issues": penalties if penalties else ["Optimal Conditions"]
        })

    # Sort crops by score in descending order
    ranked_crops.sort(key=lambda x: x["score"], reverse=True)
    return ranked_crops
```

```python
import json

sample_soil = {
    "n": 290,
    "p": 18,
    "k": 210,
    "ph": 6.2,
    "organic_matter": 0.6
}

crop_results = calculate_crop_suitability(sample_soil, CROP_DATABASE)
print(json.dumps(crop_results, indent=2))
```

```python
def calculate_soil_health_score(n, p, k, ph, organic_matter):
    """
    Computes a composite Soil Health Score (0–100) based on weighted parameters.
    Weights:
      - NPK Balance: 40%
      - pH Balance: 30%
      - Organic Matter: 30%
    """
    score = 0.0

    # 1. NPK Balance Sub-score (Max 40 points)
    npk_score = 0
    npk_score += 13.33 if THRESHOLDS["N"]["low"] <= n <= THRESHOLDS["N"]["high"] else 6.66
    npk_score += 13.33 if THRESHOLDS["P"]["low"] <= p <= THRESHOLDS["P"]["high"] else 6.66
    npk_score += 13.34 if THRESHOLDS["K"]["low"] <= k <= THRESHOLDS["K"]["high"] else 6.67
    score += npk_score

    # 2. pH Sub-score (Max 30 points)
    if THRESHOLDS["pH"]["acidic"] <= ph <= THRESHOLDS["pH"]["alkaline"]:
        score += 30.0
    else:
        ph_diff = min(abs(ph - THRESHOLDS["pH"]["acidic"]), abs(ph - THRESHOLDS["pH"]["alkaline"]))
        score += max(0.0, 30.0 - (ph_diff * 10))

    # 3. Organic Matter Sub-score (Max 30 points)
    if organic_matter >= THRESHOLDS["organic_matter"]["high"]:
        score += 30.0
    elif organic_matter >= THRESHOLDS["organic_matter"]["low"]:
        score += 20.0
    else:
        score += 10.0

    final_score = round(score, 2)

    # Health Rating Category
    if final_score >= 80:
        rating = "Excellent Soil Health"
    elif final_score >= 60:
        rating = "Good Soil Health"
    elif final_score >= 40:
        rating = "Fair / Moderate Soil Health"
    else:
        rating = "Poor / Degraded Soil Health"

    return {
        "soil_health_score": final_score,
        "rating": rating,
        "sub_scores": {
            "npk_balance": round(npk_score, 2),
            "ph_balance": round(min(30.0, max(0.0, 30.0 - (ph_diff * 10))) if not (THRESHOLDS["pH"]["acidic"] <= ph <= THRESHOLDS["pH"]["alkaline"]) else 30.0, 2),
            "organic_matter": round(30.0 if organic_matter >= THRESHOLDS["organic_matter"]["high"] else (20.0 if organic_matter >= THRESHOLDS["organic_matter"]["low"] else 10.0), 2)
        }
    }
```

```python
def run_week5_full_pipeline(soil_sample):
    """Integrates all Week 5 deliverables into a single analysis report."""
    n = soil_sample["n"]
    p = soil_sample["p"]
    k = soil_sample["k"]
    ph = soil_sample["ph"]
    om = soil_sample["organic_matter"]

    # Module 1: Recommendation Engine
    recommendations = generate_step1_recommendations(n, p, k, ph, om)

    # Module 2: Crop Suitability
    crop_suitability = calculate_crop_suitability(soil_sample, CROP_DATABASE)

    # Module 3: Soil Health Score
    health_score = calculate_soil_health_score(n, p, k, ph, om)

    return {
        "soil_sample_input": soil_sample,
        "soil_health": health_score,
        "recommendations": recommendations,
        "crop_suitability_ranking": crop_suitability
    }

# --- TEST PIPELINE EXECUTION ---
if __name__ == "__main__":
    sample_test_soil = {
        "n": 220,
        "p": 8,
        "k": 150,
        "ph": 5.4,
        "organic_matter": 0.40
    }

    report = run_week5_full_pipeline(sample_test_soil)
    print(json.dumps(report, indent=2))
```
